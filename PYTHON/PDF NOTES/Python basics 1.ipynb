```python
def fact(n):
    f=1
    for i in range(1,n+1,1):
        f=f*i
    return f;

print(fact(5))
```

```python
def fact(n):
    if(n==0 or n==1):
        return 1;
    else:
        return n*fact(n-1)

print(fact(5))
```

```python
s = 1

def college():
    global s
    s = s + 5
    print("College students: ", s)
    return s

def school():
    s = college()
    s = s + 10
    print("School students: ", s)
    return s

college()
school()
print("Global students: ",s)
```
