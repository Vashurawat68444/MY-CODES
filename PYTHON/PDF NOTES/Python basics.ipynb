```python
def fact(n):
    f=1
    for i in range(1,n+1,1):
        f=f*i

    return f;

a=int(input("Enter a Number: "))
print("Factorial is:", fact(a))
```

```python
def fact(n):
    if(n==0 or n==1):
        return 1;
    else:
        return n*fact(n-1)

print(fact(5))
```
